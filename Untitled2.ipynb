```python
import pandas as pd

# CSV file ko load karna
df = pd.read_csv('global_ecommerce_sales.csv')

# Data ke pehle 5 rows dekhne ke liye
print("First 5 rows of the dataset:")
display(df.head())

# Data ki details (columns aur data types) check karne ke liye
print("\nDataset Info:")
df.info()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Check missing values
print("Missing values in dataset:")
print(df.isnull().sum())

# 2. Basic statistics check karne ke liye
print("\nDataset Summary:")
display(df.describe())

# 3. Ek shandar graph banana (Jaise Sales ya Category ke hisaab se)
plt.figure(figsize=(10, 5))
# Apne dataset ke column ke mutabiq 'Category' ya koi aur column yahan likh sakte ho
if 'Category' in df.columns:
    sns.countplot(data=df, x='Category', palette='viridis')
    plt.title('Product Category Distribution')
    plt.xticks(rotation=45)
    plt.show()
else:
    print("Columns in your dataset:", df.columns.tolist())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Graph size set karna
plt.figure(figsize=(12, 6))

# Agar dataset mein 'Customer_Segment' column hai toh uska graph
if 'Customer_Segment' in df.columns:
    sns.barplot(data=df, x='Customer_Segment', y='Total_Sales', palette='Set2', ci=None)
    plt.title('Total Sales by Customer Segment', fontsize=14)
    plt.xlabel('Customer Segment', fontsize=12)
    plt.ylabel('Total Sales', fontsize=12)
    plt.show()
else:
    print("Columns available:", df.columns.tolist())
```
